## **Passo 1: leitura de Planilha**

Carregar os dados do arquivo brasileirao.xlsx usando pandas

```python
import pandas as pd

df = pd.read_excel('brasileirao.xlsx')
df.head()



```

## Passo 2: Calcular o saldo de gols (SG)

Adicionar uma nova coluna chamada "Saldo de gols" que será a diferença entre gols pró (GP) e gols contra (GC)

---



```python
df['Saldo de gols'] = df['GP'] - df['GC']

df.head()
```

# **Passo 3: Calcular a pontuação total**

Criar uma coluna chamada "Pontos" onde a pontuação será calculada da seguinte forma:



*   Vitória (V) = 3 pontos
*   Empate (E) = 1 ponto
*   Derrota (D) = 0 pontos







```python
df['Pontos'] = df['V'] * 3 + df['E']

df.head()
```

# **Passo 4: Filtrar os clubes com mais de 40 pontos**

Filtrar os clubes com mais de 40 pontos

```python
clubes_maiores_40 = df[df['Pontos'] > 40]

clubes_maiores_40
```

# **Passo 5: Visualizar os Dados**

Criar um gráfico de barras para visualizar a pontuação dos clubes


```python
df.plot(x='Clube', y='Pontos', kind='bar', title='Pontuação dos clubes no Brasileirão 2024')
```

```python
melhor_saldo = df[df['Saldo de gols'] == df['Saldo de gols'].max()]

melhor_saldo
```

```python
top_vitorias = df.sort_values(by='V', ascending=False).head(5)

top_vitorias
```

```python
tabela_final = df.sort_values(by='Pontos', ascending=False)

tabela_final
```
