### **Nível 1: Básico (URLs, Parâmetros e Cabeçalhos)**

Objetivo: Praticar a construção de requisições GET simples, envio de parâmetros e manipulação de cabeçalhos.


```python
import requests

# URL da API
url = "https://jsonplaceholder.typicode.com/posts"

resposta = requests.get(url)

print("Status:", resposta.status_code)

dados = resposta.json()

print(dados)
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 2
}

resposta = requests.get(url, params=params)

print("Status:", resposta.status_code)

if resposta.status_code == 200:
    dados = resposta.json()

    print("Quantidade de posts:", len(dados))

    for post in dados:
        print("ID:", post["id"])
        print("Título:", post["title"])
        print("-" * 50)
else:
    print("Erro na requisição:", resposta.status_code)
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

# Parâmetro para buscar apenas posts do usuário 2
params = {
    "userId": 2
}

# Cabeçalho personalizado
headers = {
    "User-Agent": "Projeto-Explorando-Dados-Web-Python"
}

# Fazendo a requisição
resposta = requests.get(
    url,
    params=params,
    headers=headers
)

print("Status:", resposta.status_code)

if resposta.status_code == 200:
    dados = resposta.json()

    print("Quantidade de posts:", len(dados))

    for post in dados:
        print("ID:", post["id"])
        print("Título:", post["title"])
        print("-" * 50)
else:
    print("Erro na requisição:", resposta.status_code)
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 2,
    "_limit": 3
}

headers = {
    "User-Agent": "Projeto-Explorando-Dados-Web-Python"
}

resposta = requests.get(
    url,
    params=params,
    headers=headers,
    timeout=10
)

print("Código de status:", resposta.status_code)
print("URL final:", resposta.url)

if resposta.status_code == 200:
    print("Requisição realizada com sucesso!")
else:
    print("Erro na requisição.")
```

### **Nível 2: Intermediário (JSON, Erros e Arquivos Binários)**

Objetivo: Lidar com formatos de dados reais, criar códigos mais robustos contra falhas e fazer download de mídias.


```python
import requests
import pandas as pd

ceps = [
    "24330000",
    "24220000",
    "20040002"
]

resultados = []

# Consultando cada CEP
for cep in ceps:
    url = f"https://viacep.com.br/ws/{cep}/json/"

    resposta = requests.get(url, timeout=10)

    if resposta.status_code == 200:
        dados = resposta.json()
        resultados.append(dados)

df = pd.DataFrame(resultados)

# Exibindo o resultado
print(df)
```

```python
import requests

def consultar_cep(cep):
    try:
        url = f"https://viacep.com.br/ws/{cep}/json/"

        resposta = requests.get(url, timeout=10)

        resposta.raise_for_status()

        dados = resposta.json()

        if dados.get("erro"):
            print(f"CEP {cep} não encontrado.")
            return None

        print(f"CEP {cep} consultado com sucesso!")
        return dados

    except requests.exceptions.HTTPError:
        print(f"Erro HTTP ao consultar o CEP {cep}.")
    except requests.exceptions.Timeout:
        print(f"A consulta do CEP {cep} demorou demais.")
    except requests.exceptions.RequestException:
        print(f"Não foi possível consultar o CEP {cep}.")

    return None

# Testando a função
resultado = consultar_cep("24020000")

if resultado:
    print(resultado)
```

```python
import requests

url = "https://picsum.photos/400/400"

try:
    resposta = requests.get(url, timeout=10)
    resposta.raise_for_status()

    # Salva o conteúdo da imagem em um arquivo .jpg
    with open("imagem_aleatoria.jpg", "wb") as arquivo:
        arquivo.write(resposta.content)

    print("Imagem baixada com sucesso!")

except requests.exceptions.RequestException:
    print("Não foi possível baixar a imagem.")
```

### **Nível 3: Avançado (Webscraping e Ética)**

Objetivo: Extrair dados de páginas HTML onde não há APIs estruturadas disponíveis, respeitando as boas práticas.


```python
import requests

url = "https://books.toscrape.com/robots.txt"

try:
    resposta = requests.get(url, timeout=10)
    resposta.raise_for_status()

    print("Status:", resposta.status_code)
    print("\nConteúdo do robots.txt:\n")
    print(resposta.text)

except requests.exceptions.RequestException as erro:
    print("Não foi possível acessar o robots.txt.")
    print("Erro:", erro)
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com/"

try:
    resposta = requests.get(url, timeout=10)
    resposta.raise_for_status()

    # Cria o objeto BeautifulSoup
    soup = BeautifulSoup(resposta.text, "html.parser")

    # Localiza os 5 primeiros livros
    livros = soup.select("article.product_pod")[:5]

    for livro in livros:
        # Extrai o título
        titulo = livro.find("h3").find("a")["title"]

        # Extrai o preço
        preco = livro.find("p", class_="price_color").text

        print("Título:", titulo)
        print("Preço:", preco)
        print("-" * 50)

except requests.exceptions.RequestException as erro:
    print("Erro ao acessar o site:", erro)
```
