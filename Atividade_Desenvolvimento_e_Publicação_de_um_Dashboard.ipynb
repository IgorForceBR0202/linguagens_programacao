```python
import pandas as pd

dados = {
    "produto": [
        "Notebook",
        "Mouse",
        "Teclado",
        "Monitor",
        "Headset",
        "Impressora"
    ],
    "categoria": [
        "Eletrônicos",
        "Acessórios",
        "Acessórios",
        "Eletrônicos",
        "Acessórios",
        "Informática"
    ],
    "quantidade": [
        10,
        25,
        15,
        8,
        20,
        5
    ],
    "valor": [
        5000,
        50,
        150,
        1200,
        250,
        900
    ]
}

df = pd.DataFrame(dados)

df.to_csv("vendas.csv", index=False)

print("vendas.csv criado com sucesso!")
print(df)
```

```python
%%writefile meu_dashboard.py

import streamlit as st
import pandas as pd

# CONFIGURAÇÃO DA PÁGINA

st.set_page_config(
    page_title="Dashboard de Vendas",
    page_icon="📊",
    layout="wide"
)

# TÍTULO

st.title("📊 Dashboard de Vendas")

# CARREGAMENTO DOS DADOS

@st.cache_data
def carregar_dados():
    dados = pd.read_csv("vendas.csv")
    return dados


dados = carregar_dados()

# SIDEBAR - FILTROS

st.sidebar.title("Filtros")

# Lista de categorias disponíveis
lista_de_categorias = dados["categoria"].unique()

# Multiselect de categorias
categorias_selecionadas = st.sidebar.multiselect(
    "Selecione as Categorias",
    options=lista_de_categorias
)

# FILTRAGEM DOS DADOS

if categorias_selecionadas:

    dados_filtrados = dados[
        dados["categoria"].isin(categorias_selecionadas)
    ]

else:

    dados_filtrados = dados

# RESULTADOS

st.subheader("Dados de Vendas")

st.dataframe(
    dados_filtrados,
    use_container_width=True
)
```
